Text Preprocessing

```python
#lower text
df['message'].str.lower()
df.head()
```

```python
#remove Punctuations
import re 
df['message'] = df['message'].apply(lambda x: re.sub(r'[^\w\s]', '', x))
```

Tokenization

```python
df.head()
```

```python
from nltk.tokenize import word_tokenize
df['tokens'] = df['message'].apply(word_tokenize)
df[['message', 'tokens']].head()
```

```python
#removing the stopwords
import nltk
from nltk.corpus import stopwords
nltk.download('stopwords')
stop_words = set(stopwords.words('english'))
df['tokens'] = df['tokens'].apply(lambda x: [word for word in x if word not in stop_words])
```

```python
df['tokens'].head()
```

```python
#stemming and Lemmatization
from nltk.stem import PorterStemmer
ps = PorterStemmer()

df['tokens'] = df['tokens'].apply(lambda words: [ps.stem(word) for word in words])

df.head()
```

```python
#Rejoining Tokens to text
df['processed_message'] = df['tokens'].apply(lambda x: " ".join(x))
df[['tokens', 'processed_message']].head()
```

```python
#save the cleaned df
df.to_csv('cleaned_spam_data.csv', index=False)

print("cleaned data saved as 'cleaned_spam_data.csv'")
```
